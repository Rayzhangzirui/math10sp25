```python
from math import sqrt
class Vector():
    dim = 2
    def __init__(self, input_x, input_y):
        self.x = input_x
        self.y = input_y

    def length(self):
        l = sqrt(self.x**2+self.y**2)
        return l
    def scale(self, c):
        self.x = self.x * c
        self.y = self.y * c
    def __pr__(self):
        return f'({self.x}, {self.y})'

    def add(self, other):
        x = self.x + other.x
        y = self.y + other.y
        return Vector(x,y)

    # def __add__(self, other):
    #     x = self.x + other.x
    #     y = self.y + other.y
    #     return Vector(x,y)
        
```
